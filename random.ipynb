# data analysis 

```python
import numpy as np
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

```python
# define path
data_path = 'dataset/brain_tumor_dataset.npz'
```

```python
ds = np.load(data_path)
```

```python
ds
```

```python
X_train, y_train, X_test, y_test = ds['X_train'], ds['y_train'], ds['X_test'], ds['y_test']
X_train
```

```python
X_train[0].shape
```

# Model design and training.

```python

import tensorflow as  tf 
from  tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
```

```python
X_train = X_train/255.0
X_test = X_test/255.0
```

```python

X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)
```

```python

cnn = keras.Sequential([
    layers.Conv2D(filters=32 ,kernel_size=(3,3), activation="relu", input_shape = (150,150,3)),
    layers.MaxPool2D((2,2)),
    
    layers.Conv2D(filters=64 ,kernel_size=(3,3), activation="relu"),
    layers.MaxPool2D((2,2)),
    
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')

])

cnn.compile(optimizer= 'adam', 
            loss = 'sparse_categorical_crossentropy',
            metrics=['accuracy']
            )
early_stopping = EarlyStopping(monitor='val_loss', patience=5,    restore_best_weights=True
)

model_checkpoint = ModelCheckpoint(
    'checkpoints/best_model.h5',       # File to save the best model
    monitor='val_loss',    # Monitor validation loss
    save_best_only=True
)

history =cnn.fit(X_train,y_train,epochs= 10, validation_data=(X_val, y_val), batch_size=32,
                 callbacks=[early_stopping, model_checkpoint])
```

```python
plt.plot(history.history['accuracy'], label='accuracy')
plt.plot(history.history['val_accuracy'], label = 'val_accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.ylim([0.5, 1])
plt.legend(loc='lower right')

test_loss, test_acc = cnn.evaluate(X_val,  y_val, verbose=2)
```
